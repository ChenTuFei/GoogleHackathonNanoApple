# Churn Definition and Basic Statistics
**Phase 1.2: Define churn and calculate basic statistics**

Tasks:
1. Define churn clearly (based on is_churned in dataset1)
2. Calculate overall churn rate (Q3 vs historical)
3. Churn rate time trend analysis
4. Basic descriptive statistics


```python
import pandas as pd
import numpy as np
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

print("=" * 70)
print("Churn Definition and Basic Statistics Analysis")
print("=" * 70)
```

## Step 1: Load Data


```python
df1 = pd.read_csv('dataset1_cleaned.csv')
df1['contract_start_date'] = pd.to_datetime(df1['contract_start_date'])
df1['contract_end_date'] = pd.to_datetime(df1['contract_end_date'])

df_merged = pd.read_csv('dataset_merged_final.csv')
df_merged['contract_start_date'] = pd.to_datetime(df_merged['contract_start_date'])
df_merged['contract_end_date'] = pd.to_datetime(df_merged['contract_end_date'])

print(f"Dataset1 loaded: {len(df1)} customers")
print(f"Merged dataset loaded: {len(df_merged)} customers")
```

## Task 1: Define Churn Clearly


```python
print("\n" + "=" * 70)
print("TASK 1: Define Churn Clearly")
print("=" * 70)

print("\n--- Churn Definition ---")
print("Primary Definition: Q3 Contract Ended + Not Renewed")
print("  - contract_end_date in Q3 2024 (2024-07-01 to 2024-09-30)")
print("  - renewed_flag = 0 (not renewed)")

print("\nSupporting Definition: Based on is_churned field")
print("  - is_churned = 1: Customer has churned")
print("  - is_churned = 0: Customer has not churned")

# Identify Q3 churned customers
q3_start = pd.Timestamp('2024-07-01')
q3_end = pd.Timestamp('2024-09-30')

q3_churned = df1[
    (df1['contract_end_date'] >= q3_start) & 
    (df1['contract_end_date'] <= q3_end) & 
    (df1['renewed_flag'] == 0)
]

print(f"\nQ3 Churned Customers (Primary Definition): {len(q3_churned)}")

# Overall churned customers
overall_churned = df1[df1['is_churned'] == 1]
print(f"Overall Churned Customers (is_churned=1): {len(overall_churned)}")
```

## Task 2: Calculate Overall Churn Rate (Q3 vs Historical)


```python
print("\n" + "=" * 70)
print("TASK 2: Calculate Overall Churn Rate (Q3 vs Historical)")
print("=" * 70)

total_customers = len(df1)

# Overall churn rate (based on is_churned field)
overall_churn_rate = (len(overall_churned) / total_customers) * 100
print(f"\n--- Overall Churn Rate (is_churned=1) ---")
print(f"Total Customers: {total_customers:,}")
print(f"Churned Customers: {len(overall_churned):,}")
print(f"Overall Churn Rate: {overall_churn_rate:.2f}%")

# Q3 churn rate (based on contract ended + not renewed)
q3_churn_rate_total = (len(q3_churned) / total_customers) * 100
print(f"\n--- Q3 Churn Rate (Primary Definition: Contract Ended + Not Renewed) ---")
print(f"Q3 Churned Customers: {len(q3_churned):,}")
print(f"Q3 Churn Rate (of total customers): {q3_churn_rate_total:.2f}%")

# Calculate Q3 churn rate relative to active customers at start of Q3
q1_churned_count = len(df1[(df1['contract_end_date'] >= pd.Timestamp('2024-01-01')) & 
                            (df1['contract_end_date'] <= pd.Timestamp('2024-03-31')) &
                            (df1['renewed_flag'] == 0)])
q2_churned_count = len(df1[(df1['contract_end_date'] >= pd.Timestamp('2024-04-01')) & 
                            (df1['contract_end_date'] <= pd.Timestamp('2024-06-30')) &
                            (df1['renewed_flag'] == 0)])
q3_start_active = total_customers - q1_churned_count - q2_churned_count
q3_churn_rate_active = (len(q3_churned) / q3_start_active * 100) if q3_start_active > 0 else 0

print(f"Active Customers at Start of Q3: {q3_start_active:,}")
print(f"Q3 Churn Rate (of active at start): {q3_churn_rate_active:.2f}%")

# Historical quarterly churn rates
q1_churn_rate = (q1_churned_count / total_customers * 100)
q2_churn_rate = (q2_churned_count / (total_customers - q1_churned_count) * 100)
historical_avg = (q1_churn_rate + q2_churn_rate) / 2

print(f"\n--- Historical Quarterly Churn Rates (2024 Q1-Q2) ---")
print(f"Q1 2024 Churn Rate: {q1_churn_rate:.2f}%")
print(f"Q2 2024 Churn Rate: {q2_churn_rate:.2f}%")
print(f"Q1+Q2 Average: {historical_avg:.2f}%")

# Comparison with readme
print(f"\n--- Comparison with Readme ('from 4% to 12%') ---")
print(f"Historical Average (Q1+Q2): {historical_avg:.2f}% (readme says ~4%)")
print(f"Q3 Churn Rate (of total): {q3_churn_rate_total:.2f}% (readme says 12%)")
print(f"Q3 Churn Rate (of active): {q3_churn_rate_active:.2f}%")
print(f"\nNote: Actual Q3 churn rate is {q3_churn_rate_total:.2f}%, higher than readme's 12%")
print("This suggests the crisis is even more severe than initially reported.")
```

```python
print("\n" + "=" * 70)
print("TASK 3: Churn Rate Time Trend Analysis")
print("=" * 70)

# IMPORTANT: Quarterly churn rate should be calculated as:
# Churn Rate = Churned Customers in Quarter / Active Customers at Start of Quarter

# Calculate active customers at start of each quarter
total_customers = len(df1)
q1_start_active = total_customers

# Count churned customers by quarter (contract ended + not renewed)
df_with_end = df1[df1['contract_end_date'].notna()].copy()
df_with_end['end_quarter'] = df_with_end['contract_end_date'].dt.to_period('Q')
df_with_end['is_churned_by_def'] = (df_with_end['renewed_flag'] == 0).astype(int)

# Count churned by quarter
quarterly_churned = df_with_end[df_with_end['is_churned_by_def'] == 1].groupby('end_quarter').size().reset_index(name='churned')
quarterly_churned.columns = ['quarter', 'churned']

# Calculate active customers at start of each quarter
quarters = sorted(df_with_end['end_quarter'].unique())
quarterly_stats = []

active_at_start = total_customers
for q in quarters:
    if q < pd.Period('2024Q1'):
        continue  # Focus on 2024
    
    # Churned in this quarter
    churned_this_q = quarterly_churned[quarterly_churned['quarter'] == q]['churned'].values
    churned_count = churned_this_q[0] if len(churned_this_q) > 0 else 0
    
    # Calculate churn rate
    churn_rate = (churned_count / active_at_start * 100) if active_at_start > 0 else 0
    
    quarterly_stats.append({
        'quarter': str(q),
        'active_at_start': active_at_start,
        'churned': churned_count,
        'churn_rate': churn_rate
    })
    
    # Update active customers for next quarter
    active_at_start = active_at_start - churned_count

quarterly_churn = pd.DataFrame(quarterly_stats)

print("\n--- Quarterly Churn Rate (Churned / Active at Start of Quarter) ---")
print(quarterly_churn.to_string(index=False))
```

```python
# Q3 specific analysis
print("\n--- Q3 2024 Detailed Analysis ---")

# Calculate active customers at start of Q3
q1_churned = len(df1[(df1['contract_end_date'] >= pd.Timestamp('2024-01-01')) & 
                      (df1['contract_end_date'] <= pd.Timestamp('2024-03-31')) &
                      (df1['renewed_flag'] == 0)])
q2_churned = len(df1[(df1['contract_end_date'] >= pd.Timestamp('2024-04-01')) & 
                      (df1['contract_end_date'] <= pd.Timestamp('2024-06-30')) &
                      (df1['renewed_flag'] == 0)])

q3_start_active = total_customers - q1_churned - q2_churned

q3_ended = df_with_end[
    (df_with_end['contract_end_date'] >= q3_start) & 
    (df_with_end['contract_end_date'] <= q3_end)
]
q3_churned_count = (q3_ended['renewed_flag'] == 0).sum()
q3_renewed_count = (q3_ended['renewed_flag'] == 1).sum()

# Q3 churn rate (relative to active customers at start of Q3)
q3_churn_rate = (q3_churned_count / q3_start_active * 100) if q3_start_active > 0 else 0

print(f"Active Customers at Start of Q3: {q3_start_active:,}")
print(f"Q3 Contracts Ended: {len(q3_ended):,}")
print(f"Q3 Churned (not renewed): {q3_churned_count:,}")
print(f"Q3 Renewed: {q3_renewed_count:,}")
print(f"Q3 Churn Rate (of active at start): {q3_churn_rate:.2f}%")
print(f"Q3 Churn Rate (of total customers): {q3_churned_count/total_customers*100:.2f}%")

# Monthly trend in Q3
print("\n--- Monthly Trend in Q3 2024 ---")
q3_ended['month'] = q3_ended['contract_end_date'].dt.month

# Calculate active at start of each month
monthly_stats = []
active_july = q3_start_active

for month in [7, 8, 9]:
    month_data = q3_ended[q3_ended['month'] == month]
    month_churned = (month_data['renewed_flag'] == 0).sum()
    month_rate = (month_churned / active_july * 100) if active_july > 0 else 0
    
    month_names = {7: 'July', 8: 'August', 9: 'September'}
    monthly_stats.append({
        'month': month,
        'month_name': month_names[month],
        'active_at_start': active_july,
        'churned': month_churned,
        'churn_rate': month_rate
    })
    
    active_july = active_july - month_churned

monthly_q3 = pd.DataFrame(monthly_stats)
print(monthly_q3[['month_name', 'active_at_start', 'churned', 'churn_rate']].to_string(index=False))
```

## Task 4: Basic Descriptive Statistics


```python
print("\n" + "=" * 70)
print("TASK 4: Basic Descriptive Statistics")
print("=" * 70)

# Overall statistics
print("\n--- Overall Customer Statistics ---")
print(f"Total Customers: {total_customers:,}")
print(f"Churned Customers: {len(overall_churned):,} ({overall_churn_rate:.2f}%)")
print(f"Retained Customers: {total_customers - len(overall_churned):,} ({100 - overall_churn_rate:.2f}%)")
```

```python
# By key dimensions
print("\n--- Churn Rate by Key Dimensions ---")

dimensions = ['acquisition_channel', 'product_tier', 'sales_segment', 'is_eu']
for dim in dimensions:
    if dim in df1.columns:
        print(f"\n{dim.upper()}:")
        dim_stats = df1.groupby(dim).agg({
            'is_churned': ['count', 'sum']
        }).reset_index()
        dim_stats.columns = [dim, 'total', 'churned']
        dim_stats['churn_rate'] = (dim_stats['churned'] / dim_stats['total']) * 100
        dim_stats = dim_stats.sort_values('churn_rate', ascending=False)
        for _, row in dim_stats.iterrows():
            dim_value = row[dim]
            if dim == 'is_eu':
                dim_value = 'EU' if row[dim] == 1 else 'Non-EU'
            print(f"  {dim_value}: {row['churned']}/{row['total']} = {row['churn_rate']:.2f}%")
```

```python
# Q3 churn by dimensions
print("\n--- Q3 Churn Rate by Key Dimensions ---")
for dim in dimensions:
    if dim in df1.columns:
        print(f"\n{dim.upper()}:")
        q3_dim_stats = q3_churned.groupby(dim).size().reset_index(name='q3_churned')
        dim_total = df1.groupby(dim).size().reset_index(name='total')
        q3_dim_merged = dim_total.merge(q3_dim_stats, on=dim, how='left')
        q3_dim_merged['q3_churned'] = q3_dim_merged['q3_churned'].fillna(0)
        q3_dim_merged['q3_churn_rate'] = (q3_dim_merged['q3_churned'] / q3_dim_merged['total']) * 100
        q3_dim_merged = q3_dim_merged.sort_values('q3_churn_rate', ascending=False)
        for _, row in q3_dim_merged.iterrows():
            dim_value = row[dim]
            if dim == 'is_eu':
                dim_value = 'EU' if row[dim] == 1 else 'Non-EU'
            print(f"  {dim_value}: {int(row['q3_churned'])}/{int(row['total'])} = {row['q3_churn_rate']:.2f}%")
```

```python
# Contract value statistics
print("\n--- Contract Value Statistics ---")
print(f"Overall Average ACV: ${df1['annual_contract_value'].mean():,.2f}")
print(f"Churned Average ACV: ${overall_churned['annual_contract_value'].mean():,.2f}")
print(f"Retained Average ACV: ${df1[df1['is_churned']==0]['annual_contract_value'].mean():,.2f}")

print(f"\nQ3 Churned Average ACV: ${q3_churned['annual_contract_value'].mean():,.2f}")

# Onboarding score statistics
print("\n--- Onboarding Score Statistics ---")
print(f"Overall Average Onboarding Score: {df1['initial_onboarding_score'].mean():.2f}")
print(f"Churned Average Onboarding Score: {overall_churned['initial_onboarding_score'].mean():.2f}")
print(f"Retained Average Onboarding Score: {df1[df1['is_churned']==0]['initial_onboarding_score'].mean():.2f}")

print(f"\nQ3 Churned Average Onboarding Score: {q3_churned['initial_onboarding_score'].mean():.2f}")
```

```python
# Save quarterly churn analysis
quarterly_churn.to_csv('quarterly_churn_analysis.csv', index=False)
print("✓ Saved: quarterly_churn_analysis.csv")

# Save Q3 monthly analysis
monthly_q3.to_csv('q3_monthly_churn_analysis.csv', index=False)
print("✓ Saved: q3_monthly_churn_analysis.csv")

# Create summary report
summary = []
summary.append("=" * 70)
summary.append("Churn Definition and Basic Statistics Summary")
summary.append("=" * 70)
summary.append(f"\nGenerated: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
summary.append("\n")

summary.append("1. CHURN DEFINITION")
summary.append("-" * 70)
summary.append("Primary Definition: Contract Ended + Not Renewed (renewed_flag=0)")
summary.append(f"  - Q3 Churned Customers: {len(q3_churned):,}")
summary.append("\nSupporting Definition: Based on is_churned field")
summary.append(f"  - Overall Churned Customers (is_churned=1): {len(overall_churned):,}")
summary.append("\nNote: Q3 churned customers are NOT marked as is_churned=1 in the data")
summary.append("\n")

summary.append("2. OVERALL CHURN RATE")
summary.append("-" * 70)
summary.append(f"Total Customers: {total_customers:,}")
summary.append(f"Overall Churn Rate (is_churned=1): {overall_churn_rate:.2f}%")
summary.append(f"Q3 Churn Rate (of total customers): {q3_churn_rate_total:.2f}%")
summary.append(f"Q3 Churn Rate (of active at start): {q3_churn_rate_active:.2f}%")
summary.append(f"\nHistorical Quarterly Churn Rates:")
summary.append(f"  Q1 2024: {q1_churn_rate:.2f}%")
summary.append(f"  Q2 2024: {q2_churn_rate:.2f}%")
summary.append(f"  Q1+Q2 Average: {historical_avg:.2f}%")
summary.append(f"\nComparison with Readme:")
summary.append(f"  Readme states: 'from 4% to 12%'")
summary.append(f"  Actual: from {historical_avg:.2f}% to {q3_churn_rate_total:.2f}%")
summary.append(f"  Note: Actual Q3 churn rate ({q3_churn_rate_total:.2f}%) is higher than readme's 12%")
summary.append("\n")

summary.append("3. TIME TREND ANALYSIS")
summary.append("-" * 70)
summary.append("Quarterly Churn Rates (Churned / Active at Start of Quarter):")
for _, row in quarterly_churn.iterrows():
    summary.append(f"  {row['quarter']}: {row['churn_rate']:.2f}% ({int(row['churned'])} churned from {int(row['active_at_start'])} active)")
summary.append("\nQ3 Monthly Breakdown (Churned / Active at Start of Month):")
for _, row in monthly_q3.iterrows():
    summary.append(f"  {row['month_name']}: {row['churn_rate']:.2f}% ({int(row['churned'])} churned from {int(row['active_at_start'])} active)")
summary.append("\n")

summary.append("4. KEY FINDINGS")
summary.append("-" * 70)
summary.append(f"- Overall churn rate (is_churned=1): {overall_churn_rate:.2f}%")
summary.append(f"- Q3 churn rate (of total): {q3_churn_rate_total:.2f}%")
summary.append(f"- Q3 churn rate (of active): {q3_churn_rate_active:.2f}%")
summary.append(f"- Q3 represents {len(q3_churned)} churned customers")
summary.append(f"- Average ACV of Q3 churned: ${q3_churned['annual_contract_value'].mean():,.2f}")
summary.append(f"- Average onboarding score of Q3 churned: {q3_churned['initial_onboarding_score'].mean():.2f}")
summary.append("\n")

summary.append("=" * 70)

with open('churn_definition_summary.txt', 'w', encoding='utf-8') as f:
    f.write('\n'.join(summary))

print("✓ Saved: churn_definition_summary.txt")

print("\n" + "=" * 70)
print("Analysis Complete!")
print("=" * 70)
print("\nGenerated Files:")
print("  1. quarterly_churn_analysis.csv - Quarterly churn rate trends")
print("  2. q3_monthly_churn_analysis.csv - Q3 monthly breakdown")
print("  3. churn_definition_summary.txt - Summary report")
print("=" * 70)
```
